## Question 1: Water Jug Problem using BFS

**Question:** Refer to the question statement in the companion Word document.

````python
from collections import deque

def water_jug(cap_a, cap_b, target):
    start = (0, 0)
    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        (a, b), path = queue.popleft()
        if a == target:
            return path

        next_states = [
            (cap_a, b), (a, cap_b), (0, b), (a, 0),
            (a - min(a, cap_b - b), b + min(a, cap_b - b)),
            (a + min(b, cap_a - a), b - min(b, cap_a - a))
        ]
        for state in next_states:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [state]))
    return None

path = water_jug(4, 3, 2)
print("State sequence:")
for state in path or []:
    print(state)
````

**Sample output**

```text
State sequence:
(0, 0)
(4, 0)
(1, 3)
(1, 0)
(0, 1)
(4, 1)
(2, 3)
```

Note: Each state is (amount in 4-litre jug, amount in 3-litre jug). The first solution returned depends on the order of operations.

## Question 2: 8-Puzzle using BFS

**Question:** Refer to the question statement in the companion Word document.

````python
from collections import deque

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)
goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

def neighbours(state):
    z = state.index(0)
    r, c = divmod(z, 3)
    result = []
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            nz = nr * 3 + nc
            s = list(state)
            s[z], s[nz] = s[nz], s[z]
            result.append(tuple(s))
    return result

def bfs(start, goal):
    queue = deque([start])
    parent = {start: None}
    while queue:
        state = queue.popleft()
        if state == goal:
            path = []
            while state is not None:
                path.append(state)
                state = parent[state]
            return path[::-1]
        for nxt in neighbours(state):
            if nxt not in parent:
                parent[nxt] = state
                queue.append(nxt)
    return None

path = bfs(start, goal)
if path:
    for step, state in enumerate(path):
        print("Step", step)
        for i in range(0, 9, 3):
            print(state[i:i+3])
else:
    print("No solution found")
````

```text
(4, 0, 6)
(7, 5, 8)
Step 1
(1, 2, 3)
(4, 5, 6)
(7, 0, 8)
Step 2
(1, 2, 3)
(4, 5, 6)
(7, 8, 0)
```

Note: BFS is suitable for this small example. For harder 8-Puzzle instances, BFS may use substantial memory.

## Question 3: BFS on a Graph

**Question:** Refer to the question statement in the companion Word document.

````python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def bfs(graph, start, goal):
    queue = deque([start])
    visited = {start}
    parent = {start: None}
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)
        if node == goal:
            break
        for nxt in graph[node]:
            if nxt not in visited:
                visited.add(nxt)
                parent[nxt] = node
                queue.append(nxt)

    path = []
    if goal in visited:
        cur = goal
        while cur is not None:
            path.append(cur)
            cur = parent[cur]
        path.reverse()
    return order, path

order, path = bfs(graph, 'A', 'F')
print("BFS traversal:", order)
print("Path to goal:", path if path else "Goal not reachable")
````

## Question 4: DFS on a Graph

**Question:** Refer to the question statement in the companion Word document.

````python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def dfs(graph, start, goal):
    visited = set()
    order = []
    path = []

    def visit(node):
        visited.add(node)
        order.append(node)
        path.append(node)
        if node == goal:
            return True
        for nxt in graph[node]:
            if nxt not in visited and visit(nxt):
                return True
        path.pop()
        return False

    found = visit(start)
    return order, path if found else []

order, path = dfs(graph, 'A', 'F')
print("DFS traversal:", order)
print("Path to goal:", path if path else "Goal not reachable")
````

## Question 5: Best-First Search

**Question:** Refer to the question statement in the companion Word document.

````python
import heapq

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}
h = {'A': 5, 'B': 4, 'C': 2, 'D': 5, 'E': 1, 'F': 0}

def best_first(start, goal):
    frontier = [(h[start], start, [start])]
    visited = set()
    order = []
    while frontier:
        _, node, path = heapq.heappop(frontier)
        if node in visited:
            continue
        visited.add(node)
        order.append(node)
        if node == goal:
            return order, path
        for nxt in graph[node]:
            if nxt not in visited:
                heapq.heappush(frontier, (h[nxt], nxt, path + [nxt]))
    return order, []

order, path = best_first('A', 'F')
print("Expansion order:", order)
print("Path:", path)
````

**Sample output**

```text
Expansion order: ['A', 'C', 'F']
Path: ['A', 'C', 'F']
```

Note: Greedy Best-First Search chooses the node with the lowest heuristic value; it does not guarantee an optimal-cost path.

## Question 6: A* Search using f(n) = g(n) + h(n)

**Question:** Refer to the question statement in the companion Word document.

````python
import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 5), ('E', 2)],
    'C': [('E', 1)],
    'D': [('G', 3)],
    'E': [('G', 5)],
    'G': []
}
h = {'A': 7, 'B': 6, 'C': 4, 'D': 3, 'E': 4, 'G': 0}

def astar(start, goal):
    pq = [(h[start], 0, start, [start])]
    best_g = {start: 0}
    while pq:
        f, g, node, path = heapq.heappop(pq)
        if node == goal:
            return path, g
        if g != best_g.get(node):
            continue
        for nxt, cost in graph[node]:
            ng = g + cost
            if ng < best_g.get(nxt, float('inf')):
                best_g[nxt] = ng
                heapq.heappush(pq, (ng + h[nxt], ng, nxt, path + [nxt]))
    return [], None

path, cost = astar('A', 'G')
print("Optimal path:", path)
print("Total cost:", cost)
````

## Question 7: A* Search with Path Cost Comparison

**Question:** Refer to the question statement in the companion Word document.

````python
import heapq

graph = {
    'S': [('A', 2), ('B', 5)],
    'A': [('G', 6), ('B', 1)],
    'B': [('G', 3)],
    'G': []
}
h = {'S': 5, 'A': 4, 'B': 2, 'G': 0}

def astar(start, goal):
    pq = [(h[start], 0, start, [start])]
    best = {start: 0}
    while pq:
        _, cost, node, path = heapq.heappop(pq)
        if node == goal:
            return path, cost
        if cost != best[node]:
            continue
        for nxt, w in graph[node]:
            nc = cost + w
            if nc < best.get(nxt, float('inf')):
                best[nxt] = nc
                heapq.heappush(pq, (nc + h[nxt], nc, nxt, path + [nxt]))
    return [], None

path, cost = astar('S', 'G')
print("A* path:", path, "Cost:", cost)
print("Alternative route S-B-G cost:", 5 + 3)
````

## Question 8: Minimax Algorithm

**Question:** Refer to the question statement in the companion Word document.

````python
def minimax(tree, maximizing):
    if isinstance(tree, int):
        return tree
    values = [minimax(child, not maximizing) for child in tree]
    return max(values) if maximizing else min(values)

game_tree = [
    [3, 5, 2],
    [9, 1, 4]
]
print("Best value for MAX:", minimax(game_tree, True))
````

**Sample output**

```text
Best value for MAX: 2
```

Note: Each internal level alternates between MAX and MIN. The example tree has a MAX root and MIN children.

## Question 9: Alpha-Beta Pruning

**Question:** Refer to the question statement in the companion Word document.

````python
def alphabeta(node, depth, alpha, beta, maximizing, count):
    if isinstance(node, int):
        count[0] += 1
        return node
    if maximizing:
        value = float('-inf')
        for child in node:
            value = max(value, alphabeta(child, depth+1, alpha, beta, False, count))
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value
    else:
        value = float('inf')
        for child in node:
            value = min(value, alphabeta(child, depth+1, alpha, beta, True, count))
            beta = min(beta, value)
            if alpha >= beta:
                break
        return value

tree = [[3, 5, 6], [9, 1, 2]]
count = [0]
value = alphabeta(tree, 0, float('-inf'), float('inf'), True, count)
print("Minimax value:", value)
print("Leaf nodes evaluated:", count[0])
````

**Sample output**

```text
Minimax value: 3
Leaf nodes evaluated: 5
```

Note: Pruning efficiency depends on the order of children in the game tree.

## Question 11: Forward Chaining

**Question:** Refer to the question statement in the companion Word document.

````python
facts = {"has_feathers", "lays_eggs"}
rules = [
    ({"has_feathers"}, "is_bird"),
    ({"is_bird", "lays_eggs"}, "may_be_egg_laying_bird"),
    ({"is_bird"}, "has_wings")
]

changed = True
while changed:
    changed = False
    for conditions, conclusion in rules:
        if conditions.issubset(facts) and conclusion not in facts:
            facts.add(conclusion)
            print("Inferred:", conclusion)
            changed = True

print("Final facts:", sorted(facts))
````

## Question 12: Backward Chaining

**Question:** Refer to the question statement in the companion Word document.

````python
facts = {"has_feathers", "lays_eggs"}
rules = {
    "is_bird": [{"has_feathers"}],
    "may_be_egg_laying_bird": [{"is_bird", "lays_eggs"}],
    "has_wings": [{"is_bird"}]
}

def prove(goal, seen=None):
    if goal in facts:
        return True
    if seen is None:
        seen = set()
    if goal in seen:
        return False
    seen.add(goal)
    for conditions in rules.get(goal, []):
        if all(prove(condition, seen.copy()) for condition in conditions):
            return True
    return False

goal = "may_be_egg_laying_bird"
print("Goal:", goal)
print("Proved:", prove(goal))
````

```text
Goal: may_be_egg_laying_bird
Proved: True
```

## Question 17: BFS: Goal Reachability

**Question:** Refer to the question statement in the companion Word document.

````python
from collections import deque

graph = {'A': ['B', 'C'], 'B': ['D', 'E'],
         'C': ['F'], 'D': [], 'E': [], 'F': []}

def bfs(start, goal):
    queue = deque([start])
    visited = {start}
    order = []
    parent = {start: None}
    while queue:
        node = queue.popleft()
        order.append(node)
        if node == goal:
            break
        for nxt in graph[node]:
            if nxt not in visited:
                visited.add(nxt)
                parent[nxt] = node
                queue.append(nxt)
    path = []
    if goal in visited:
        node = goal
        while node is not None:
            path.append(node)
            node = parent[node]
        path.reverse()
    return order, path

order, path = bfs('A', 'E')
print("Traversal:", order)
print("Reachable:", bool(path))
print("Path:", path)
````

**Sample output**

```text
Traversal: ['A', 'B', 'C', 'D', 'E']
Reachable: True
Path: ['A', 'B', 'E']
```

## Question 23: A* Search with Cost Comparison

**Question:** Refer to the question statement in the companion Word document.

````python
import heapq

graph = {
    'S': [('A', 2), ('B', 5)],
    'A': [('B', 1), ('G', 6)],
    'B': [('G', 3)],
    'G': []
}
h = {'S': 5, 'A': 4, 'B': 2, 'G': 0}

def astar(start, goal):
    pq = [(h[start], 0, start, [start])]
    best = {start: 0}
    while pq:
        _, g, node, path = heapq.heappop(pq)
        if g != best.get(node):
            continue
        if node == goal:
            return path, g
        for nxt, cost in graph[node]:
            ng = g + cost
            if ng < best.get(nxt, float('inf')):
                best[nxt] = ng
                heapq.heappush(pq, (ng + h[nxt], ng, nxt, path + [nxt]))
    return [], None

path, cost = astar('S', 'G')
print("Path:", path)
print("A* cost:", cost)
print("Alternative S-B-G cost:", 8)
````

**Sample output**

## Question 24: Hill Climbing for Route Optimization

**Question:** Refer to the question statement in the companion Word document.

````python
# Each route is represented by a label and its cost.
neighbours = {
    "Route A": [("Route B", 18), ("Route C", 14)],
    "Route B": [("Route A", 22), ("Route D", 12)],
    "Route C": [("Route A", 20), ("Route D", 10)],
    "Route D": [("Route B", 15), ("Route C", 13)]
}
costs = {"Route A": 20, "Route B": 18, "Route C": 14, "Route D": 10}

current = "Route A"
print("Start:", current, "cost =", costs[current])

while True:
    better = [(route, cost) for route, cost in neighbours[current]
              if cost < costs[current]]
    if not better:
        break
    current, _ = min(better, key=lambda item: item[1])
    print("Move to:", current, "cost =", costs[current])

print("Final route:", current)
print("Final cost:", costs[current])
````

## Question 25: Water Jug Problem using State-Space Search

**Question:** Refer to the question statement in the companion Word document.

````python
from collections import deque

def solve():
    cap_a, cap_b, target = 4, 3, 2
    start = (0, 0)
    queue = deque([start])
    parent = {start: None}

    while queue:
        a, b = queue.popleft()
        if a == target:
            goal = (a, b)
            path = []
            while goal is not None:
                path.append(goal)
                goal = parent[goal]
            return path[::-1]

        pour_ab = min(a, cap_b - b)
        pour_ba = min(b, cap_a - a)
        states = [
            (cap_a, b), (a, cap_b), (0, b), (a, 0),
            (a - pour_ab, b + pour_ab),
            (a + pour_ba, b - pour_ba)
        ]
        for state in states:
            if state not in parent:
                parent[state] = (a, b)
                queue.append(state)
    return []

for state in solve():
    print(state)
````

**Sample output**

```text
(0, 0)
(4, 0)
(1, 3)
(1, 0)
(0, 1)
(4, 1)
```

## Quick revision: main algorithms

| Algorithm | Main idea |
|---|---|
| BFS | Queue; explores level by level |
| DFS | Recursion/stack; explores deeply before backtracking |
| Greedy Best-First Search | Selects the frontier node with the lowest heuristic value |
| A* Search | Uses `f(n) = g(n) + h(n)` |
| Minimax | Alternates between MAX and MIN levels |
| Alpha-Beta Pruning | Skips branches that cannot affect the Minimax result |
| Hill Climbing | Moves to a better neighbouring state |
| Forward Chaining | Uses known facts to infer new facts |
| Backward Chaining | Starts with a goal and tries to prove it |
